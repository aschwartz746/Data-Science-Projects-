```python
#Importing the libraries
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import sklearn
%matplotlib inline
!pip install mglearn
import mglearn
from sklearn.model_selection import train_test_split
from sklearn.datasets import load_wine
from sklearn.pipeline import make_pipeline
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.preprocessing import PolynomialFeatures
import seaborn as sns
```

```python
#Load the data and check the first 5 rows
wine = load_wine()
wine_df = pd.DataFrame(wine.data, columns=wine.feature_names)
wine_df.head()
```

```python
#Check dataset information
wine_df.info()
```

```python
#Check descriptive stats
wine_df.describe()
```

```python
#Check the target
y = wine.target
y
```

```python
#Split the training and testing data on the wine dataset
X_train, X_test, y_train, y_test = train_test_split(
    wine.data, wine.target, random_state=0)
```

```python
#Ridge Regression Model

#Import Ridge

from sklearn.linear_model import Ridge

#Use the default setting, alpha = 1
ridge = Ridge(alpha=1).fit(X_train, y_train)
print("Training set score: {:.2f}".format(ridge.score(X_train, y_train)))
print("Test set score: {:.2f}".format(ridge.score(X_test, y_test)))
```

```python
#Lasso Regression model

#Import Lasso
from sklearn.linear_model import Lasso

#Use the default setting, alpha = 1
lasso = Lasso(alpha=1).fit(X_train, y_train)
print("Training set score: {:.2f}".format(lasso.score(X_train, y_train)))
print("Test set score: {:.2f}".format(lasso.score(X_test, y_test)))
print("Number of features used:", np.sum(lasso.coef_ != 0))
```

```python
#Make a pipline with StandardScalar(), PolynomialFeatures(), and Ridge()
pipe = make_pipeline(
    StandardScaler(),
    PolynomialFeatures(),
    Ridge())
```

```python
#Define the parameter grid as poly_degree: 1, 2, 3; regressor_alpha: 0.01, 0.1, 1, 10, 100 - this is for the Pipeline Ridge
param_grid = {'polynomialfeatures__degree': [1, 2, 3],
              'ridge__alpha': [0.01, 0.1, 1, 10,  100]}
```

```python
#Runing the GridSearchCV based on the Pipeline Ridge

#Importing the GridSearchCV
from sklearn.model_selection import GridSearchCV

#Running the GridSearchCV
gridr = GridSearchCV(pipe, param_grid=param_grid, cv=5, n_jobs=-1)
gridr.fit(X_train, y_train)
```

```python
#Checking the ridge parameters
print("Best parameters: {}".format(gridr.best_params_))
```

```python
#Checking the Pipeline Ridge test score
print("Test-set score (Pipeline Ridge): {:.2f}".format(gridr.score(X_test, y_test)))
```

```python
#Make a pipline with StandardScalar(), PolynomialFeatures(), and Lasso()
pipe = make_pipeline(
    StandardScaler(),
    PolynomialFeatures(),
    Lasso())
```

```python
#Define the parameter grid as poly_degree: 1, 2, 3; regressor_alpha: 0.01, 0.1, 1, 10, 100 - this is for the Pipeline Lasso
param_grid = {'polynomialfeatures__degree': [1, 2, 3],
              'lasso__alpha': [0.01, 0.1, 1, 10,  100]}
```

```python
#Running the GridSearchCV based on the Pipeline Lasso
gridl = GridSearchCV(pipe, param_grid=param_grid, cv=5, n_jobs=-1)
gridl.fit(X_train, y_train)
```

```python
#Checking the Lasso Parameters
print("Best parameters: {}".format(gridl.best_params_))
```

```python
#Checking the Pipeline Lasso test score
print("Test-set score (Pipeline Lasso): {:.2f}".format(gridl.score(X_test, y_test)))
```
